```python
divida_inicial = float(input("Digite o valor da dívida inicial: "))
taxa = float(input("Digite a taxa de juros mensal (%): "))
pagamento_mensal = float(input("Digite o pagamento mensal: "))

juros_primeiro_mes = divida_inicial * (taxa / 100)

if pagamento_mensal <= juros_primeiro_mes:
    print("Pagamento insuficiente.")
    print("A dívida nunca será paga.")
else:
    mes = 0
    divida = divida_inicial
    total_pago = 0.0

    while divida > 0:
        juros = divida * (taxa / 100)
        divida = divida + juros
        mes = mes + 1

        if divida >= pagamento_mensal:
            pago_neste_mes = pagamento_mensal
            divida = divida - pagamento_mensal
        else:
            pago_neste_mes = divida
            divida = 0.0

        total_pago = total_pago + pago_neste_mes

    total_juros = total_pago - divida_inicial

    print("Meses:", mes)
    print("Total pago: R$", total_pago)
    print("Total de juros: R$", total_juros)
```

```python
desposito = float(input("Digite o valor do deposito: "))
taxa = float(input("Digite a taxa de rendimento mensal: (%) "))

mes = 1
saldo = desposito
while mes <= 24:
  rendimento = saldo * (taxa / 100)
  saldo = saldo + rendimento

  print("Mês:", mes)
  print("Saldo: R$", saldo)

  total_juros = saldo - desposito

  mes = mes + 1

print("Total de juros: R$", total_juros)
```
