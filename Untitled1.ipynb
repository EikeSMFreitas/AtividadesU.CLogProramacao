```python
A = float(input("Valor de A: "))
B = float(input("Valor de B: "))

R = A + B

print(f"Resultado da soma R:{R} ")
```

```python
preco_mercadoria = float(input("Preço da mercadoria: "))
percentual_desconto = float(input("Percentual de desconto: "))
desconto = preco_mercadoria * (percentual_desconto / 100)
preco_final = preco_mercadoria - desconto

print(f"Seu desconto foi de: {desconto}% O preço final da mercadoria é R${preco_final:.2f}")
#
```

```python
distancia = float(input("Distancia (km): "))
velocidade = float(input("Velocidae (km/h): "))

tempo = distancia/velocidade

print(f"Tempo estimado: {tempo:.2f} horas")
```

```python
num_01 = int(input("Digite um número: "))
num_02 = int(input("Digite outro número: "))
num_03 = int(input("Digite mais um número: "))


# Alternativa usando condicionais aninhados

if num_01 > num_02 and num_01 > num_03:
    maior = num_01
elif num_02 > num_01 and num_02 > num_03:
    maior = num_02
else:
    maior = num_03
if num_01 < num_02 and num_01 < num_03:
    menor = num_01
elif num_02 < num_01 and num_02 < num_03:
    menor = num_02
else:
    menor = num_03
print(f"O maior número é {maior} e o menor é {menor}.")

```
