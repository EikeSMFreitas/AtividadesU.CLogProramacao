```python
A = float(input("Valor de A: "))
B = float(input("Valor de B: "))

R = A + B

print(f"Resultado da soma R:{R} ")
```

```python
preco_mercadoria = float(input("Preço da mercadoria: "))
percentual_desconto = float(input("Percentual de desconto: "))
desconto = preco_mercadoria * (percentual_desconto / 100)
preco_final = preco_mercadoria - desconto

print(f"Seu desconto foi de: {desconto}% O preço final da mercadoria é R${preco_final:.2f}")
#
```
